```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("a) Reversed:", arr[::-1])
```

```python
array1 = np.array([[1, 2, 3], [2, 4, 5], [1, 2, 3]])
print("b) Flatten (flatten):", array1.flatten())
print("   Flatten (ravel):", array1.ravel())
```

```python
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[1, 2], [3, 4]])
print("c) Arrays equal:", np.array_equal(arr1, arr2))
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
val_x = np.bincount(x).argmax()
val_y = np.bincount(y).argmax()
print("d) x: most frequent =", val_x, "at", np.where(x == val_x)[0])
print("   y: most frequent =", val_y, "at", np.where(y == val_y)[0])
```

```python
gfg = np.matrix('[4, 1, 9; 12, 3, 1; 4, 5, 6]')
print("e) Total sum:", gfg.sum())
print("   Row-wise:", gfg.sum(axis=1))
print("   Column-wise:", gfg.sum(axis=0))
```

```python
n = np.array([[55, 25, 15],[30, 44, 2],[11, 45, 77]])
print("f) Diagonal sum:", np.trace(n))
print("   Eigen values:", np.linalg.eigvals(n))
print("   Inverse:\n", np.linalg.inv(n))
print("   Determinant:", np.linalg.det(n))
```

```python
p1 = np.array([[1, 2], [2, 3]])
q1 = np.array([[4, 5], [6, 7]])
print("g) i. Dot product:\n", np.dot(p1, q1))
```

```python
p2 = np.array([[1, 2], [2, 3], [4, 5]])
q2 = np.array([[4, 5, 1], [6, 7, 2]])
print("   ii. Dot product:\n", np.dot(p2, q2))
```

```python
x = np.array([[2, 3, 4], [3, 2, 9]])
y = np.array([[1, 5, 0], [5, 10, 3]])
print("h) Inner:\n", np.inner(x, y))
print("   Outer:\n", np.outer(x, y))
print("   Cartesian product:\n", np.array(np.meshgrid(x.flatten(), y.flatten())).T.reshape(-1, 2))
```

```python


array = np.array([[1, -2, 3], [-4, 5, -6]])

print(np.abs(array))

print(np.percentile(array, 25))
print(np.percentile(array, 50))
print(np.percentile(array, 75))

print(np.percentile(array, 25, axis=0))
print(np.percentile(array, 50, axis=0))
print(np.percentile(array, 75, axis=0))

print(np.percentile(array, 25, axis=1))
print(np.percentile(array, 50, axis=1))
print(np.percentile(array, 75, axis=1))

print(np.mean(array))
print(np.median(array))
print(np.std(array))

print(np.mean(array, axis=0))
print(np.median(array, axis=0))
print(np.std(array, axis=0))

print(np.mean(array, axis=1))
print(np.median(array, axis=1))
print(np.std(array, axis=1))

a = np.array([-1.8, -1.6, -0.5, 0.5, 1.6, 1.8, 3.0])

print(np.floor(a))
print(np.ceil(a))
print(np.trunc(a))
print(np.round(a))
```

```python
from PIL import Image
import numpy as np

def img_to_array(path):
    img = Image.open(path)
    arr = np.array(img)
    if len(arr.shape) == 2:
        np.savetxt("grayscale_image.txt", arr, fmt='%d')
        print(f"{path} is a Grayscale image. Saved as 'grayscale_image.txt'")
    else:
        reshaped = arr.reshape(-1, arr.shape[-1])
        np.savetxt("rgb_image.txt", reshaped, fmt='%d')
        print(f"{path} is an RGB image. Saved as 'rgb_image.txt'")

img_to_array("gray_image.jpg")
img_to_array("color_image.jpg")

print("\n--- Loaded from grayscale_image.txt ---")
gray = np.loadtxt("grayscale_image.txt", dtype=int)
print(gray)

print("\n--- Loaded from rgb_image.txt ---")
rgb = np.loadtxt("rgb_image.txt", dtype=int).reshape(-1, 3)
print(rgb)
```
